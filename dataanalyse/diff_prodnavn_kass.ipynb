Her undersøges afvigelsen i 5-6 år.

```python
df = datasæt["T-shirt"]
df = df[df["Produkt - Produkt"] == "09385 T-Shirt Hvid B/P M Arla Tryk"]
df["Kassationsårsag (ui)"].value_counts()
```

Dette en stor procentdel af det samlede antal kasseringer pga misligholdt, $\frac{957}{3596}*100=26.61\%$, for denne arla trøje der finder sted sammentidig

Undersøger nu 2-3 år spiken

```python
df = datasæt["T-shirt"]
df = df[df["Produkt - Produkt"] == "09385 T-Shirt Hvid B/P M Arla Tryk"]
df = df[(df["Dage i cirkulation"] >= 365.25 * 2) & (df["Dage i cirkulation"] <= 365.25*3)]
df["Kassationsårsag (ui)"].value_counts()
```
